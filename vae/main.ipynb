```python
import torch
import torch.nn as nn

# ---------- inplace=False ----------
x = torch.tensor([-1.0, 2.0, -3.0])
print("原始 x:", x)

relu_false = nn.ReLU(inplace=False)
y = relu_false(x)

print("inplace=False -> x:", x)   # 没变
print("inplace=False -> y:", y)   # 新张量
print("x 和 y 是同一对象吗:", x is y)   # False

print("-" * 40)

# ---------- inplace=True ----------
x2 = torch.tensor([-1.0, 2.0, -3.0])
print("原始 x2:", x2)

relu_true = nn.ReLU(inplace=True)
z = relu_true(x2)

print("inplace=True -> x2:", x2)  # 被改掉了
print("inplace=True -> z:", z)    # 和 x2 是同一个
print("x2 和 z 是同一对象吗:", x2 is z)  # True
```

```python
import torch
import torch.nn as nn

# N=2, C=3, H=2, W=2
x = torch.randn(2, 3, 2, 2)

bn = nn.BatchNorm2d(3, affine=False)
bn.train()          # 训练模式，用当前 batch 的统计量
y = bn(x)

# 手动按通道计算
for c in range(3):
    channel_data = x[:, c, :, :]          # 形状 (N, H, W)
    mean = channel_data.mean()
    var  = channel_data.var(unbiased=False)

    manual = (channel_data - mean) / torch.sqrt(var + 1e-5)
    print(f"通道 {c}:")
    print("  BN 输出:  ", y[:, c, :, :].flatten())
    print("  手动计算:", manual.flatten())
    print()
```
